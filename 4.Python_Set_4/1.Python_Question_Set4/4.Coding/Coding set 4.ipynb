```python
#3. Find All Armstrong Numbers Between 1 and 1000  153 = 1**3 + 5**3 + 3 **3
#Question: Write a Python program to print all Armstrong numbers between 1 and 1000.
    
num = 153
ln=len(str(num))
res = 0
for i in str(num):
    res += pow(int(i),ln)
    
if (num == res):
    print("arm")
```

```python
def armsornot(num):
    ln=len(str(num))
    res = 0
    for i in str(num):
        res += pow(int(i),ln)
    
    return res

#     if (num == res):
#         print("arm")
```

```python
lst=[]
for i in range(1,1000):
    ans = armsornot(i)
    if(i == ans):
        lst.append(i)
        
print(lst)
```

```python
#4.Count the Frequency of Each Character in a String
#Question: Write a Python program to count the frequency of each character in "programming".

word = "programming"
temlist = {}
for ch in word:
    if ch in temlist:
        temlist[ch] += 1
    else:
        temlist[ch] = 1

print(temlist)        
```

```python
#5. Generate a List of Squares of Numbers from 1 to 10
#Question: Write a Python program to generate a list containing squares of numbers from 1 to 10.

temlist=[]
res=0
for num in range(1, 10):
    res = num ** 2
    temlist.append(res)
print(temlist)
```
