## Question 1: Multi-Agent Role Interaction (Reviewer ➝ Editor ➝ Finalizer)
Task: Implement a three-agent pipeline where each agent performs a unique role. Pass the message correctly to avoid format errors.

```python
from autogen import ConversableAgent

config = {
        "model": "gemini-flash-latest",
        "api_key": "AQ.Ab8RN6JU2Etd3OS98AH7sXtAE_wYmXQ9jGfoSO0N27x_NEoa6g",  # Add key
        "api_type": "google",
}

reviewer = ConversableAgent(name="Reviewer", llm_config=config)
editor = ConversableAgent(name="Editor", llm_config=config)
Finalizer = ConversableAgent(name="Finalizer", llm_config=config)

abstract = "This research explores the use of transformers in low-resource languages."

# Step 1: Reviewer
review = reviewer.generate_reply(
    messages=[{"role": "user", "content": abstract}]
)

review_text = review if isinstance(review, str) else review.get("content", "")

# Step 2: Editor 
edited = editor.generate_reply(
    messages=[{"role": "user", "content": review_text}]    #pass string only
)

edited_text = edited if isinstance(edited, str) else edited.get("content", "")

# Step 3: Finalizer
summary = Finalizer.generate_reply(
    messages=[{"role": "user", "content": edited_text}]
)

summary_text = summary if isinstance(summary, str) else summary.get("content", "")

print("\n=== REVIEW ===\n", review_text)
print("\n=== EDITED ===\n", edited_text)
print("\n=== FINAL ===\n", summary_text)
```

## Question 2: Language Translator ➝ Culture Rephraser
Task: Use one agent to translate from English to French, then another to culturally rephrase the French text.

```python
from autogen import ConversableAgent

# LLM Configuration
config = {
    "config_list": [
        {
        "model": "gemini-flash-latest",
        "api_key": "AQ.Ab8RN6JU2Etd3OS98AH7sXtAE_wYmXQ9jGfoSO0N27x_NEoa6g",  # Add key
        "api_type": "google",
        }
    ]
}

# Create Agents
translator = ConversableAgent(
    name="Translator",
    llm_config=config
)

culture_rephraser = ConversableAgent(
    name="CultureRephraser",
    llm_config=config
)

# Input text
english_text = "Good morning! Have a great day."

# Agent 1: Translate English → French
translation = translator.generate_reply(
    messages=[
        {
            "role": "user",
            "content": f"Translate the following English text into French:\n\n{english_text}"
        }
    ]
)

# Extract text safely
if isinstance(translation, dict):
    french_text = translation["content"]
else:
    french_text = translation

print("French Translation:")
print(french_text)

# Agent 2: Culturally rephrase the French text
rephrased = culture_rephraser.generate_reply(
    messages=[
        {
            "role": "user",
            "content": (
                "Rewrite the following French text so that it sounds "
                "natural and culturally appropriate for a native French speaker:\n\n"
                f"{french_text}"
            )
        }
    ]
)

# Extract text safely
if isinstance(rephrased, dict):
    final_text = rephrased["content"]
else:
    final_text = rephrased

print("\nCulturally Rephrased French:")
print(final_text)
```

## Question 3: Code Generator ➝ Code Reviewer ➝ Bug  Fixer
Task: Have Agent A generate Python code, Agent B review it, and Agent C suggest bug fixes.


```python
from autogen import ConversableAgent

# LLM Configuration
config = {
    "config_list": [
        {
        "model": "gemini-flash-latest",
        "api_key": "AQ.Ab8RN6JU2Etd3OS98AH7sXtAE_wYmXQ9jGfoSO0N27x_NEoa6g",  # Add key
        "api_type": "google",
        }
    ]
}

# Create Agents
generator = ConversableAgent(name="CodeGenerator", llm_config=config)

reviewer = ConversableAgent(name="CodeReviewer", llm_config=config)

bug_fixer = ConversableAgent(name="BugFixer",llm_config=config)


# Helper function to extract text
def get_text(response):
    if isinstance(response, dict):
        return response.get("content", "")
    return response


#Generate code
generated_code = get_text(
    generator.generate_reply(
        messages=[
            {
                "role": "user",
                "content": (
                    "Write a Python function to check whether a number is prime."
                )
            }
        ]
    )
)

print("\n========== Generated Code ==========\n")
print(generated_code)


# Review the code
review = get_text(
    reviewer.generate_reply(
        messages=[
            {
                "role": "user",
                "content": (
                    f"Review the following Python code and identify any bugs, "
                    f"logic errors, or improvements:\n\n{generated_code}"
                )
            }
        ]
    )
)

print("\n========== Code Review ==========\n")
print(review)


# Fix the code
fixed_code = get_text(
    bug_fixer.generate_reply(
        messages=[
            {
                "role": "user",
                "content": (
                    f"Based on the following review:\n\n{review}\n\n"
                    f"Rewrite and fix the original Python code:\n\n{generated_code}"
                )
            }
        ]
    )
)

print("\n========== Fixed Code ==========\n")
print(fixed_code)
```

## Question 4: Email Response Automation using Multiple Agents
Task:
    Create a 3-agent flow using ConversableAgent and Gemini:
    
●	ReaderAgent: Extracts main point from an email

●	ResponderAgent: Drafts a reply

●	FormatterAgent: Formats it as a proper email


```python
from autogen import ConversableAgent

# LLM Configuration
config = {
    "config_list": [
        {
            "model": "gemini-flash-latest",
            "api_key": "AQ.Ab8RN6JU2Etd3OS98AH7sXtAE_wYmXQ9jGfoSO0N27x_NEoa6g",  # Add key
            "api_type": "google",
        }
    ]
}

# Create Agents
reader_agent = ConversableAgent(name="ReaderAgent",llm_config=config)

responder_agent = ConversableAgent(name="ResponderAgent",llm_config=config)

formatter_agent = ConversableAgent(name="FormatterAgent",llm_config=config)

# Helper function
def get_text(response):
    if isinstance(response, dict):
        return response.get("content", "")
    return response

# Sample Email
email = """
Subject: Project Deadline

Hi Team,

The client has requested that we submit the final project by Friday instead of next Monday.
Please confirm whether you can complete your assigned tasks before the new deadline.

Regards,
Project Manager
"""


#-----Reader Agent-----------

main_point = get_text(
    reader_agent.generate_reply(
        messages=[
            {
                "role": "user",
                "content": f"""
Read the following email and extract its main point in 2-3 sentences.

Email:
{email}
"""
            }
        ]
    )
)

print("\n========== Main Point ==========\n")
print(main_point)


#-----Responder Agent-----------

reply = get_text(
    responder_agent.generate_reply(
        messages=[
            {
                "role": "user",
                "content": f"""
Based on the following email summary, draft a professional reply confirming that all assigned tasks will be completed before Friday.

Summary:
{main_point}
"""
            }
        ]
    )
)

print("\n------ Draft Reply --------\n")
print(reply)


# ---------Formatter Agent------------

formatted_email = get_text(
    formatter_agent.generate_reply(
        messages=[
            {
                "role": "user",
                "content": f"""
Format the following response as a professional email.
Include:
- Subject
- Greeting
- Body
- Closing

Draft:
{reply}
"""
            }
        ]
    )
)

print("\n------- Final Email --------\n")
print(formatted_email)
```

##  Question 5: Meeting Transcript → Action Points
  Task: Create a 3-agent pipeline to convert a meeting transcript into actionable tasks.
      
●	ListenerAgent: Extracts main discussion points

●	ActionAgent: Converts them into to-do tasks

●	DeadlineAgent: Assigns timelines to each task


```python
from autogen import ConversableAgent

# LLM Configuration
config = {
    "config_list": [
        {
            "model": "gemini-flash-latest",
            "api_key": "AQ.Ab8RN6JU2Etd3OS98AH7sXtAE_wYmXQ9jGfoSO0N27x_NEoa6g",  # Add key
            "api_type": "google",
        }
    ]
}

# Create Agents
listener_agent = ConversableAgent(
    name="ListenerAgent",
    llm_config=config
)

action_agent = ConversableAgent(
    name="ActionAgent",
    llm_config=config
)

deadline_agent = ConversableAgent(
    name="DeadlineAgent",
    llm_config=config
)

# Helper function
def get_text(response):
    if isinstance(response, dict):
        return response.get("content", "")
    return response

# Sample Meeting Transcript
transcript = """
Project Manager: We need to complete the user authentication module this week.

Developer: I'll finish the login and registration APIs by Wednesday.

UI Designer: I'll update the login and signup screens before Thursday.

QA Engineer: Testing can begin once development is complete.

Project Manager: Let's deploy the application by next Monday after successful testing.
"""


#------Listener Agent--------------

discussion_points = get_text(
    listener_agent.generate_reply(
        messages=[
            {
                "role": "user",
                "content": f"""
Read the following meeting transcript and summarize the key discussion points.

Transcript:
{transcript}
"""
            }
        ]
    )
)

print("\n========== Discussion Points ==========\n")
print(discussion_points)


#------Action Agent-------------

tasks = get_text(
    action_agent.generate_reply(
        messages=[
            {
                "role": "user",
                "content": f"""
Based on these discussion points, create a clear to-do list with bullet points.

Discussion Points:
{discussion_points}
"""
            }
        ]
    )
)

print("\n========== Action Items ==========\n")
print(tasks)


#----------Deadline Agent------------

final_plan = get_text(
    deadline_agent.generate_reply(
        messages=[
            {
                "role": "user",
                "content": f"""
Assign realistic deadlines to each task based on the meeting notes.

Tasks:
{tasks}

Return the result as a table with:
- Task
- Deadline
"""
            }
        ]
    )
)

print("\n========== Final Action Plan ==========\n")
print(final_plan)
```
