## Solution

```python
from autogen import ConversableAgent

config = {
    "config_list": [
        {
            "model": "gemini-3.1-flash-lite",
            "api_key": "AQ.Ab8RN6JsNxzcyyFLtYZLPavrRlfG_4bEnGCX1Ssq1BqyE1a6Nw",
            "api_type": "google",
        }
    ]
}

reviewer = ConversableAgent(name="Reviewer", llm_config=config)
translator = ConversableAgent(name="Translator", llm_config=config)
assigner = ConversableAgent(name="TaskAssigner", llm_config=config)

def extract(resp):
    if isinstance(resp, dict):
        return resp.get("content", "")
    return resp

# 1. Reviewer
review = extract(
    reviewer.generate_reply(
        messages=[{
            "role": "user",
            "content": "Review this sentence: The cat are on the table."
        }]
    )
)

print("Reviewer:", review)

# 2. Translator
translated = extract(
    translator.generate_reply(
        messages=[{
            "role": "user",
            "content": review
        }]
    )
)

print("Translator:", translated)

# 3. Task Assigner
task = extract(
    assigner.generate_reply(
        messages=[{
            "role": "user",
            "content": translated
        }]
    )
)

print("TaskAssigner:", task)
```
