```python
import pandas as pd  

data = {'Name': ['Alice', 'Bob', 'Charlie'], 
        'Age': [25, 30, 'Thirty-Five'],  # Error: 35 is written in words, this will give error
        'Gender': ['F', 'M', 'M']}  

df = pd.DataFrame(data)

df['Age']=df['Age'].replace('Thirty-Five', 35) #Convert the word Thirty-Five to numberic 35

average_age = df['Age'].mean()

print(f"Average Age: {average_age}")
```

```python
import numpy as np
from scipy.stats import zscore

data = np.array([1, 2, 3, 100, 5, 6, 7, 8, 9])
median = np.median(data)

#data = [median if abs(zscore(x)) > 2 else x for x in data] #Error direct using of Zcore

#Below is correct code
zc = zscore(data)
#print(median, zc)
data = [median if abs(z) > 2 else x 
        for x, z in zip(data,zc)]
data
```
