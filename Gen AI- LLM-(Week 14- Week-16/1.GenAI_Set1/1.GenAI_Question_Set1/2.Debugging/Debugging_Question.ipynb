## Solution

```python
#tokenizer.encode is for single sentence. Use tokenizer() to handle batches

from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
text = ["What is RAG?", "Explain Generative AI."]
tokens = tokenizer(text, padding=True, truncation=True)
tokens
```

## Question 3

```python
from transformers import pipeline

summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

text = "Python."

summary = summarizer(text)  

print(summary)
```

## Solution

```python
# summarizer assigned with single word. Assign summarizer with longer text and set max and min length


from transformers import pipeline

summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

text = """
CNN.com will feature iReporter photos in a weekly Travel Snapshots gallery. 
Visit CNN.com/Travel each week for a new gallery of snapshots from around the world. 
Please submit your best shots of the world for next week's gallery. 
Send your best photos to jennifer.smith@mailonline.co.uk.
"""

summary = summarizer(text, max_length=30, min_length=5)  

print(summary)
```
