## Question 1: 
## You are trying to use the T5 model to perform English-to-French translation. The model expects tokenized input.

```python
from transformers import T5Tokenizer, T5ForConditionalGeneration

token = T5Tokenizer.from_pretrained("t5-small")
model = T5ForConditionalGeneration.from_pretrained("t5-small")

text="translate English to French: How are you?"

inputs = token(text, return_tensors="pt")
outputs = model.generate(**inputs)

translate = token.decode(outputs[0], skip_special_tokens=True)

print("French Translation:", translate)
```

## Question 2:

## You wrote a RAG-like prompt flow but skipped the vector retrieval step. Complete it.


```python
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np

# Load embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Documents
documents = [
    "AI is transforming industries.",
    "FAISS enables efficient vector search.",
    "Generative AI creates human-like responses."
]

# Convert documents into embeddings
doc_embeddings = model.encode(documents)

# Convert to float32
doc_embeddings = np.array(doc_embeddings).astype("float32")

# Create FAISS index
dimension = doc_embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)

# Add embeddings to index
index.add(doc_embeddings)

# User query
query = "How does vector search work?"

# Query embedding
query_embedding = model.encode([query]).astype("float32")

# Search nearest document
distances, indices = index.search(query_embedding, k=1)

# Retrieve best matching document
retrieved_doc = documents[indices[0][0]]

print("Retrieved Document:")
print(retrieved_doc)
```

## Question 3:

## You are trying to pass plain text to a model without tokenizing. Fix it to generate correct output using the Hugging Face T5 model.


```python
from transformers import T5Tokenizer, T5ForConditionalGeneration

# Load tokenizer and model
tokenizer = T5Tokenizer.from_pretrained("t5-small")
model = T5ForConditionalGeneration.from_pretrained("t5-small")

# Plain text input
text = "translate English to French: Hello, how are you?"

# Tokenize the input text
inputs = tokenizer(text, return_tensors="pt")

# Generate output tokens
outputs = model.generate(**inputs)

# Decode tokens into readable text
result = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("French Translation:", result)
```

## Question 4:

## What is the correct code to summarize a paragraph using a pre-trained BART model from Hugging Face?


```python
from sentence_transformers import SentenceTransformer, util


embed_model = SentenceTransformer('all-MiniLM-L6-v2')


documents = [
    "Artificial Intelligence is transforming healthcare and finance.",
    "FAISS is widely used for vector similarity search.",
    "Deep learning powers modern generative AI systems."
]


doc_embeddings = embed_model.encode(documents, convert_to_tensor=True)


query = "How is AI used in industries?"


query_embedding = embed_model.encode(query, convert_to_tensor=True)


scores = util.cos_sim(query_embedding, doc_embeddings)


best_doc_index = scores.argmax().item()
retrieved_doc = documents[best_doc_index]

print("Retrieved Document:")
print(retrieved_doc)
```

## Question 5:

## You want to generate a small sentence using a GenAI model. Use the Hugging Face pipeline to do it with a tiny model. What’s the correct code?

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="gpt2"
)

result = generator(
    "Artificial Intelligence is",
    max_length=30,
    num_return_sequences=1,
    temperature=0.7
)

print(result[0]['generated_text'])
```
