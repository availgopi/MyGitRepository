<h2> 7)Test the Analysis of Variance between etest_p and mba_p at signifance
level 5%.(Make decision using Hypothesis Testing) </h2>
<h4>P < 0.05 </BR>
    H0=There is no siginificant difference </BR>
    H1=There is siginificant difference between etest_p and mba_p </h4> 

```python
from scipy.stats import stats
etestp= dataset['etest_p']
mbap=dataset['mba_p']
result = stats.f_oneway(etestp,mbap)
print(f"P-value: {result.pvalue:.6f}")
if(result.pvalue < 0.05):
    print("Reject the Null Hypothesis")
    print("There is a statistically significant difference between the two groups.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is no statistically significant difference between the two groups.")
# print(f"F-statistic: {result.statistic:.6f}")
# print(f"P-value: {result.pvalue:.6f}")
```

<h2> 8)Test the similarity between the degree_t(Sci&Tech) and specialisation(
Mkt&HR) with respect to salary at significance level of 5%.(Make
decision using Hypothesis Testing) </h2>
<h3> P < 0.05 </BR>
    Reject H0: There is a significant difference in mean salary </h3> 

```python
from scipy.stats import ttest_rel,stats
sci_tech = dataset[dataset['degree_t'] == 'Sci&Tech']['salary']
Mkt_Hr = dataset[dataset['specialisation'] == 'Mkt&HR']['salary']
result = stats.f_oneway(sci_tech, Mkt_Hr)
print(f"P-value: {result.pvalue:.6f}")
if(result.pvalue < 0.05):
    print("Reject H0: There is a significant difference in mean salary")
else:
    print("Fail to reject H0: No significant difference in mean salary")
```

<h2> 10)What is the probability Density Function of the salary range from
700000 to 900000? </h2>

```python
def getpdf(dataset,startrange, endrange):
    import seaborn as sns
    from scipy.stats import norm
    import matplotlib.pyplot as plt
    ax = sns.distplot(dataset, kde=True, kde_kws={'color':'blue'}, color='Green')
    plt.axvline(startrange, color='Red')
    plt.axvline(endrange, color='Red')

    dt = pd.to_numeric(dataset, errors="coerce")
    mn = dt.mean()
    sd = dt.std()
    print(f"Mean={mn:.2f}, StandardDeviation={sd:.2f}")                  

    probabilitiy = norm.cdf(endrange, mn, sd) - norm.cdf(startrange, mn, sd)

    # z_score = norm(mn, sd)
    # values = [value for value in range(700000,900000)]
    # prob = [z_score.pdf(value) for value in values]
    # probabilities = sum(prob)
    print("The salary between range({},{}):{}".format(startrange, endrange, probabilitiy))
```

```python
getpdf(dataset["salary"], 700000, 900000)
```

<h2> 11)Test the similarity between the degree_t(Sci&Tech)with respect to
etest_p and mba_p at significance level of 5%.(Make decision using Hypothesis
Testing) </h2>
<h4>P < 0.05 </BR>
    H0-There is no similarity
</h4>

```python
from scipy.stats import ttest_rel, stats
etest = dataset[dataset['degree_t'] == 'Sci&Tech']['etest_p']
mba = dataset[dataset['degree_t'] == 'Sci&Tech']['mba_p']
result = stats.f_oneway(etest, mba)
print(f"{result.pvalue:6f}")
if(result.pvalue < 0.05):
    print("Reject the Null Hypothesis")
    print("There is a statistically significant difference between the two groups.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is no statistically significant difference between the two groups.")
```
